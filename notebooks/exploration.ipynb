## Paso 1: Importar Liberias

```python
import pandas as pd
import numpy as np
import requests
import os
```

## Paso 2: Descargar los datos

```python
# URL for the Titanic dataset
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"

# Ensure the '../data' directory exists
os.makedirs('../data', exist_ok=True)  # Creates folder if it doesn't exist

try:
    # Download the dataset
    response = requests.get(url)
    response.raise_for_status()  # Check for HTTP errors

    # Save to '../data/titanic_data.csv'
    with open('../data/titanic_data.csv', 'wb') as f:
        f.write(response.content)

    print("Dataset saved successfully to ../data/titanic_data.csv")
except requests.exceptions.RequestException as e:
    print("Failed to download the dataset:", e)
except Exception as e:
    print("An error occurred:", e)
```

```python
df = pd.read_csv('../data/titanic_data.csv')
```

```python
# Select specific numerical columns
selected_columns = ['Survived', 'sqfeet', 'beds', 'baths']
numerical_columns = [col for col in selected_columns if col in df.columns]

# Compute summary statistics
statistics = df[numerical_columns].agg(
    ['mean', 'median', 'std', 'var', 'min', 'max', 'skew', 'kurt']
).T.rename(columns={
    "mean": "Mean",
    "median": "Median",
    "std": "Standard Deviation",
    "var": "Variance",
    "min": "Minimum",
    "max": "Maximum",
    "skew": "Skewness",
    "kurt": "Kurtosis"
})

# Display results
print("\nMedidas de tendencia central y dispersión:\n")
statistics
```
